```python
Advance Data Type conversion
1-List
2-Set
3-Tuple
4-Dictionary
```

```python
1 List-> set ,tuple ,dict
```

```python
lst=[1,2,3]
print(lst , type(lst))
```

```python
#set conversion
se=set(lst)
print(se,type(se))
```

```python
#tuple conversion
tp=tuple(lst)
print(tp,type(tp))
```

```python
#dict conversion
dt=dict(lst)
print(dt,type(dt))
```

```python
2 Set-> list ,tuple ,dict
```

```python
se={1,2,3,4}
print(se,type(se))
```

```python
#list conversion
li=list(se)
print(li,type(li))
```

```python
#tuple conversion
tp1=tuple(se)
print(tp1,type(tp1))
```

```python
#dict conversion
di=dict(se)
print(di,type(di))
```

```python
3  Tuple -> list ,set ,dict
```

```python
tp2=(1,2,3,4)
print(tp2,type(tp2))
```

```python
#list conversion
lst2=list(tp2)
print(lst2,type(lst2))
```

```python
#set conversion
st2=set(tp2)
print(st2,type(st2))
```

```python
#dict conversion
di=dict(tp2)
print(di,type(di))
```

```python
4  Dictionary -> list ,set ,tuple
```

```python
di={1:'mango',2:'banana'}
print(di,type(di))
```

```python
#list convertion
lst3=list(di)
print(lst3,type(lst3))
```

```python
#set convertion
st3=set(di)
print(st3,type(st3))
```

```python
#tuple convertion
tp4=tuple(di)
print(tp4,type(tp4))
```
