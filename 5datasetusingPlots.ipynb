```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# 1. Penguins 🐧

```python
# Load the dataset first (This defines 'df')
df = sns.load_dataset("penguins")
df
```

## 🐧 Histogram 

```python
plt.figure(figsize=(9,5))
sns.histplot(data=df, x='flipper_length_mm', hue='species', kde=True, multiple='layer', alpha=0.6)
plt.title('Flipper Length Distribution: Across Penguin Species')
plt.xlabel('Flipper Length (mm)')
plt.ylabel('Frequency')
plt.grid(True)
plt.tight_layout()
plt.show()
```

## This code creates a layered histogram with smooth trend lines to compare the distribution of flipper lengths across different penguin species.

## 🐧 Box Plot

```python
plt.figure(figsize=(9,5))
sns.boxplot(data=df, x='species', y='body_mass_g', hue='sex', palette='coolwarm')
plt.title('Penguin Body Mass Spread by Species and Sex')
plt.xlabel('Species')
plt.ylabel('Body Mass (g)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

## This code creates a grouped box plot to compare the distribution, medians, and ranges of penguin body mass across different species, further broken down by sex

## 🐧 Scatter Plot

```python
plt.figure(figsize=(9,6))
sns.scatterplot(data=df, x='bill_length_mm', y='bill_depth_mm', hue='species', size='body_mass_g', sizes=(30, 250), palette='viridis', alpha=0.75)
plt.title('Bill Length vs Bill Depth, Sized by Body Mass')
plt.xlabel('Bill Length (mm)')
plt.ylabel('Bill Depth (mm)')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0)
plt.grid(True)
plt.tight_layout()
plt.show()
```

## This code creates a multivariate scatter plot to analyze the relationship between bill length and bill depth across species, while representing body mass through point sizes.

# 2. Titanic 🛥️

```python
# Load the dataset first (This defines 'df')
df = sns.load_dataset("titanic")
df
```

##  🛥️ Histogram 

```python
plt.figure(figsize=(9, 5))
sns.histplot(data=df, x="age", hue="survived", kde=True, multiple="layer", alpha=0.6,)
plt.title("Age Distribution: Across Titanic Passengers")
plt.xlabel("Age (Years)")
plt.ylabel("Frequency")
plt.grid(True)
plt.tight_layout()
plt.show()
```

### This code creates a layered histogram with smooth trend lines to compare the frequency and distribution of passenger ages based on whether they survived or died on the Titanic.

##  🛥️ Relplot

```python
plt.figure(figsize=(9, 5))
sns.relplot(data=df,x="age",y="fare",  hue="survived", kind="scatter",alpha=0.6,height=5,aspect=1.8)
plt.title("Age Distribution: Across Titanic Passengers", y=1.03)
plt.xlabel("Age (Years)")
plt.ylabel("Fare")
plt.grid(True)
plt.tight_layout()
plt.show()
```

### This code creates a multivariate scatter plot to analyze the relationship between passenger age and ticket fare, color-coded by survival status across Titanic passengers.

## 🛥️ Boxplot

```python
plt.figure(figsize=(9, 5))
sns.boxplot(data=df, x="class", y="fare", hue="sex", palette="coolwarm", fliersize=4)
plt.title("Ticket Price Spread broken down by Class and Sex", y=1.03)
plt.xlabel("Passenger Cabin Class")
plt.ylabel("Fare Paid ($)")
plt.ylim(0, 300)  # Caps excessive outliers to keep the grid clear
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### This code creates a grouped box plot to compare the distribution, median, and outlier ranges of ticket fares paid across different cabin classes, further broken down by passenger sex.

# 3 Diamonds 💎 

```python
# Load the dataset first (This defines 'df')
df = sns.load_dataset("diamonds")
df
```

## 💎 Violinplot 

```python
plt.figure(figsize=(9, 5))
sns.violinplot( data=df, x="cut", y="carat", hue="cut", legend=False, palette="muted", inner="box")
plt.title("Density Distribution of Carat Weight by Cut Quality", y=1.03)
plt.xlabel("Cut Quality")
plt.ylabel("Carat Weight")
plt.ylim(0, 3)  
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### This code creates a violin plot with internal box plots to compare the probability density and weight distribution of diamond carats across different cut qualities.

## 💎 Heatmap

```python
plt.figure(figsize=(9, 6))
numerical_df = df.select_dtypes(include=["float64", "int64"])
corr_matrix = numerical_df.corr()
sns.heatmap(data=corr_matrix,annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix Heatmap of Diamond Features", y=1.03)
plt.tight_layout()
plt.show()
```

### This code creates an annotated heatmap matrix to measure and visually check the strength of linear relationships between all numerical features of the diamonds.

## 💎 Countplot

```python
plt.figure(figsize=(9, 5))
clarity_order = ["I1", "SI2", "SI1", "VS2", "VS1", "VVS2", "VVS1", "IF"]
sns.countplot( data=df, x="clarity", hue="color", order=clarity_order, palette="tab10")
plt.title("Inventory Count by Clarity and Color Grade", y=1.03)
plt.xlabel("Clarity Grade")
plt.ylabel("Number of Diamonds Available")
plt.grid(True, axis="y", linestyle="--", alpha=0.5)  
plt.tight_layout()
plt.show()
```

### This code creates a grouped count plot to compare the inventory volume of diamonds across ordered clarity grades, further broken down by color categories.

# 4 Taxis 🚕 

```python
# Load the dataset first (This defines 'df')
df = sns.load_dataset("taxis")
df
```

## 🚕 Barplot

```python
plt.figure(figsize=(9, 5))
clean_boroughs = df[df["dropoff_borough"].notna()]
sns.barplot(data=clean_boroughs, x="dropoff_borough", y="distance", hue="dropoff_borough",   legend=False,palette="Set2",errorbar=None)
plt.title("Average Trip Distance by Dropoff Borough Destination", y=1.03)
plt.xlabel("Dropoff Borough")
plt.ylabel("Mean Distance (Miles)")
plt.grid(True, axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### This code creates an aggregated bar plot to compare the average trip distance driven across different dropoff borough destinations without displaying error bars.

## 🚕 Lineplot

```python
df["pickup"] = pd.to_datetime(df["pickup"])
df["pickup_date"] = df["pickup"].dt.date
df["pickup_hour"] = df["pickup"].dt.hour

plt.figure(figsize=(10, 5))
sns.lineplot( data=df, x="pickup_date", y="distance", hue="color", palette={"yellow": "#f1c40f", "green": "#2ecc71"},  linewidth=2)
plt.title("Daily Average Trip Distance: Yellow vs Green Taxis", y=1.03)
plt.xlabel("Date (March 2019)")
plt.ylabel("Average Distance (Miles)")
plt.xticks(rotation=45)  
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### This code creates a time-series line plot to analyze and compare the daily average trip distances traveled by yellow and green taxis over time.

## 🚕 Relplot

```python
hourly_fare = ( df.groupby(["pickup_hour", "color"])["total"].mean().reset_index())
sns.relplot(data=hourly_fare,x="pickup_hour",y="total",hue="color",kind="line",
            marker="o",  palette={"yellow": "#f1c40f", "green": "#2ecc71"},height=5,aspect=1.8,)
plt.title("Average Total Trip Cost Across 24 Hours", y=1.03)
plt.xlabel("Hour of the Day (0 - 23)")
plt.ylabel("Mean Total Fare ($)")
plt.xticks(range(0, 24, 2))  
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### This code creates an hourly relational line plot to compare the fluctuation of average total trip costs across a 24-hour cycle between yellow and green taxis.

# 5 Planets 🪐 

```python
# Load the dataset first (This defines 'df')
df = sns.load_dataset("planets")
df
```

## 🪐 Countplot

```python
plt.figure(figsize=(9, 5))
top_methods = df["method"].value_counts().index[:5]
filtered_df = df[df["method"].isin(top_methods)]
sns.countplot( data=filtered_df, y="method", hue="method",  order=top_methods, palette="viridis", legend=False,)
plt.title("Top 5 Exoplanet Discovery Methods by Volume", y=1.03)
plt.xlabel("Number of Planets Discovered")
plt.ylabel("Detection Method")
plt.grid(True, axis="x", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()
```

### This code creates a horizontal count plot to compare the total volume of exoplanets discovered by the top five most successful detection methods.

## 🪐 Histplot

```python
plt.figure(figsize=(9, 5))
sns.histplot(data=df,x="distance",hue="method",hue_order=["Radial Velocity", "Transit", "Imaging"], kde=True,
    multiple="layer",alpha=0.5,log_scale=True,  palette="Set1",)
plt.title("Logarithmic Distribution of Exoplanet Distances", y=1.03)
plt.xlabel("Distance to Planet (Log Scale - Light Years)")
plt.ylabel("Frequency")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

### This code creates a log-scaled layered histogram with trend lines to compare the spatial distance distributions of exoplanets discovered via Radial Velocity, Transit, and Imaging.

## 🪐 Boxplot

```python
plt.figure(figsize=(9, 5))
top_4_methods = ["Radial Velocity", "Transit", "Imaging", "Microlensing"]
df_top_4 = df[df["method"].isin(top_4_methods)]
sns.boxplot(data=df_top_4,x="method",y="orbital_period",hue="method", legend=False,palette="pastel",)
plt.yscale("log")  
plt.title("Distribution of Planet Orbital Periods by Method", y=1.03)
plt.xlabel("Discovery Method")
plt.ylabel("Orbital Period (Days - Log Scale)")
plt.grid(True, axis="y", linestyle="--", alpha=0.5, which="both")
plt.tight_layout()
plt.show()
```

### This code creates a log-scaled box plot to compare the distribution, median ranges, and outliers of planet orbital periods across the top four exoplanet discovery methods.
