```python
2)	float - > int , str , bool , complex 
```

```python
a = 12.8
print("Original Value :", a)
print("Original Type  :", type(a))
print("\nfloat to int     :", int(a))
print("Type               :", type(int(a)))
print("\nfloat to str     :", str(a))
print("Type            :", type(str(a)))
print("\nfloat to bool    :", bool(a))
print("Type            :", type(bool(a)))
print("\nfloat to complex :", complex(a))
print("Type            :", type(complex(a)))
```
