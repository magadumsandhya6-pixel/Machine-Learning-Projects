```python
# 41. Implement gradient descent to fit a line y = mx + b to noisy data.
x = np.array([1, 2, 3, 4, 5])
y = np.array([2.1, 3.9, 6.1, 8.0, 10.2])

m, b = 0.0, 0.0
lr = 0.01
epochs = 1000
n = len(x)

for _ in range(epochs):
    y_pred = m * x + b
    dm = (-2/n) * np.sum(x * (y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)
    m -= lr * dm
    b -= lr * db

print(f"Fitted line: y = {m:.2f}x + {b:.2f}")
```

```python
# 46. Build a simple image-patch extractor: given an image array and patch size, extract all non-overlapping patches.
image = np.arange(16).reshape(4, 4) # 4x4 image matrix
p_h, p_w = 2, 2 # Patch dimensions

# Native reshape/transpose tricks to split spatial planes without copies
h, w = image.shape
patches = image.reshape(h // p_h, p_h, w // p_w, p_w).swapaxes(1, 2).reshape(-1, p_h, p_w)
print("Extracted Patches:\n", patches)
```

```python
# 47. Compute cosine similarity between two sets of vectors (matrix form, no loops).
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8], [1, 0]])

dot_product = A @ B.T
norm_A = np.linalg.norm(A, axis=1, keepdims=True)
norm_B = np.linalg.norm(B, axis=1, keepdims=True)

cosine_sim = dot_product / (norm_A @ norm_B.T)
print("Cosine Similarity Matrix:\n", cosine_sim)
```

```python
# 48. Implement a basic Markov chain transition matrix and compute the stationary distribution via eigenvectors.
P = np.array([[0.7, 0.3], [0.4, 0.6]]) # Row-stochastic transition matrix

# Find left eigenvectors by computing right eigenvectors of P.T
eigenvalues, eigenvectors = np.linalg.eig(P.T)
stationary_vector = eigenvectors[:, np.isclose(eigenvalues, 1.0)].real.flatten()

# Normalize vector so probabilities equal 1
stationary_distribution = stationary_vector / np.sum(stationary_vector)
print("Stationary Distribution:", stationary_distribution)
```

```python
# 50. Vectorized PSNR computation over image batches (Batch, H, W, C) without Python loops.
def batch_psnr(img_batch1, img_batch2, max_val=255.0):
    # Compute MSE across spatial and color channels dimensions (axis=1,2,3)
    mse = np.mean((img_batch1 - img_batch2) ** 2, axis=(1, 2, 3))
    
    # Handle absolute clean-match cases to prevent division by zero errors
    mse = np.where(mse == 0, 1e-10, mse)
    
    psnr = 20 * np.log10(max_val) - 10 * np.log10(mse)
    return psnr

# Mock batches: 2 images, 32x32 resolution, 3 RGB channels
b1 = np.random.randint(0, 256, (2, 32, 32, 3)).astype(float)
b2 = b1 + np.random.normal(0, 2, (2, 32, 32, 3)) # Add minor noise

print("PSNR per image in batch:", batch_psnr(b1, b2))
```
