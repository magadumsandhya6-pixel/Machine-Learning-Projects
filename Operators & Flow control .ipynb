## Operators & Flow control 

```python
#Operators
#1.Arithmetic operator  
a = 10
b = 3

print(a + b)   # 13
print(a - b)   # 7
print(a * b)   # 30
print(a / b)   # 3.333...
print(a // b)  # 3
print(a % b)   # 1
print(a ** b)  # 1000
```

```python
#2.relational operators
a = 10
b = 5

print(a == b)   # False
print(a != b)   # True
print(a > b)    # True
print(a < b)    # False
print(a >= b)   # True
print(a <= b)   # False
```

```python
#2.relational operators
a = 10
b = 5

print(a == b)   # False
print(a != b)   # True
print(a > b)    # True
print(a < b)    # False
print(a >= b)   # True
print(a <= b)   # False
```

```python
#2. or
a = 10
b = 5

print(a > 5 or b > 10)    # True (one condition is true)
print(a < 5 or b > 10)    # False
```

```python
#3. not
a = 10

print(not (a > 5))        # False (because a > 5 is True)
print(not (a < 5))        # True
```

```python
#4.Bitwise operators
a = 5   # 0101 in binary
b = 3   # 0011 in binary

print("a & b =", a & b)   # AND
print("a | b =", a | b)   # OR
print("a ^ b =", a ^ b)   # XOR
print("~a =", ~a)         # NOT
print("a << 1 =", a << 1) # Left shift
print("a >> 1 =", a >> 1) # Right shift
```

```python
#5.assignment operator
x = 10

# =
x = 10
print("=  :", x)

# +=
x += 5
print("+= :", x)

# -=
x -= 3
print("-= :", x)

# *=
x *= 2
print("*= :", x)

# /=
x /= 4
print("/= :", x)

# //=
x //= 2
print("//= :", x)

# %=
x %= 2
print("%= :", x)

# **=
x **= 3
print("**= :", x)
```

```python
#5.Bitwise Assignment Examples
a = 5

a &= 3
print(a)  # 1

a = 5
a |= 3
print(a)  # 7

a = 5
a ^= 3
print(a)  # 6

a = 5
a <<= 1
print(a)  # 10

a = 5
a >>= 1
print(a)  # 2
```

```python
#6.special operator
#a.identify operators
x = [1, 2, 3]
y = x
z = [1, 2, 3]

print(x is y)      # True
print(x is z)      # False
print(x is not z)  # True
```

```python
#b.membership operator
numbers = [10, 20, 30, 40]

print(20 in numbers)
print(50 in numbers)

print(20 not in numbers)
print(50 not in numbers)
```

## Operator procedence
🔷 BODMAS (used in UK, India, etc.)

B – Brackets

O – Orders (powers, square roots, exponents)

D – Division

M – Multiplication

A – Addition

S – Subtraction

```python
#operator procedence
# 1. Parentheses ()
print((10 + 5) * 2)      # 30

# 2. Exponentiation **
print(2 ** 3)            # 8

# 3. Unary Operators +, -, ~
print(-5)                # -5
print(+5)                # 5
print(~5)                # -6

# 4. Multiplication, Division, Floor Division, Modulus
print(10 * 2)            # 20
print(10 / 2)            # 5.0
print(10 // 3)           # 3
print(10 % 3)            # 1

# 5. Addition, Subtraction
print(10 + 5)            # 15
print(10 - 5)            # 5

# 6. Left Shift, Right Shift
print(5 << 1)            # 10
print(5 >> 1)            # 2

# 7. Bitwise AND
print(5 & 3)             # 1

# 8. Bitwise XOR
print(5 ^ 3)             # 6

# 9. Bitwise OR
print(5 | 3)             # 7

# 10. Relational Operators
print(10 > 5)            # True
print(10 == 10)          # True

# 11. Logical NOT
print(not True)          # False

# 12. Logical AND
print(True and False)    # False

# 13. Logical OR
print(True or False)     # True
```

```python
#Important Precedence Example
result = 10 + 5 * 2 > 15 and not False
print(result)
```

```python
#complex operator precedence
z = (1 + 2j) + (3 + 4j) * 2
print(z)
```

```python
#Step-by-step:
#Step 1:8/2*3+5+(12/4)**2-6
#Step 2: (solve division inside parentheses)4*3+5+(3.0)**2-6
#Step 3: (solve exponent)4*3+5+9.0-6
#Step 4: (multiplication)12+5+9.0-6
#Step 5: (left to right addition/subtraction)17+9.0-6
#Step 6:20.0
result=8/2*3+5+(12/4)**2-6
print(result)
```

## Flow control 
Conditional - 1.if , 2.if-else, 3.if-elif-else

Iterative - 1.for ,2.while

Transfer - 1.break, 2.continue,3.pass

## 1️⃣ Conditional Statements

Used to make decisions based on conditions.

```python
#1. if
#Executes a block only if condition is True.

x = 10
if x > 5:
    print("x is greater than 5")
```

```python
age = 20
if age >= 18:
    print("Eligible to vote")
```

```python
#2. if-else
#Chooses between two blocks.

x = 3
if x > 5:
    print("Greater")
else:
    print("Smaller or equal")
```

```python
num = 15
if num % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")
```

```python
#3. if-elif-else
#Used for multiple conditions.

x = 7
if x > 10:
    print("Big")
elif x > 5:
    print("Medium")
else:
    print("Small")
```

```python
marks = 78
if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

## 2️⃣ Iterative (Looping) Statements

Used to repeat a block of code.

```python
#1. for loop
#Used when number of iterations is known.

for i in range(5):
    print(i)
```

```python
total = 0
for i in range(1, 6):
    total += i
print("Sum =", total)
```

```python
#2. while loop
#Runs while condition is True.
i = 0
while i < 7:
    print(i)
    i += 1
```

```python
num = 5
while num > 0:
    print(num)
    num -= 1
print("Blast Off!")
```
